# RiskLens: Exploratory Data Analysis

```
import os, sys
os.chdir('..')
sys.path.append('.')
import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
from src.data_loader import load_raw
from src.preprocess import preprocess
sns.set_theme(style='whitegrid')
```

## 1. Overview

```
raw = load_raw()
print('Shape:', raw.shape)
print(raw.dtypes)
print('Default rate:', round(raw['SeriousDlqin2yrs'].mean(), 4))
```

## 2. Missing values before and after cleaning

```
df = preprocess(raw)
missing_before = raw.isna().sum()
print('Before cleaning:')
print(missing_before[missing_before > 0])
print('After cleaning (total missing):', df.isna().sum().sum())
```

## 3. Default rate by key features

```
df['age_group'] = pd.cut(df['age'], [17, 25, 35, 45, 55, 65, 120])
df['util_bucket'] = pd.cut(df['utilization'], [-0.01, 0.1, 0.3, 0.6, 0.9, 10])
df['total_late'] = df[['late_30_59', 'late_60_89', 'late_90']].sum(axis=1).clip(upper=6)

fig, ax = plt.subplots(1, 3, figsize=(16, 4))
for a, col in zip(ax, ['age_group', 'util_bucket', 'total_late']):
    df.groupby(col, observed=True)['default'].mean().plot(kind='bar', ax=a)
    a.set_title(f'Default rate by {col}')
    a.set_ylabel('default rate')
plt.tight_layout()
plt.show()
```

## 4. Correlation heatmap

```
num = df.select_dtypes('number')
plt.figure(figsize=(10, 8))
sns.heatmap(num.corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation matrix')
plt.show()
```
